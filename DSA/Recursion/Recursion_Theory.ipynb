## 1. Introduction to Recursion & Call Stack

**What is Recursion?**
Recursion is simply a function calling itself to solve smaller instances of the same problem, until it hits a trivial condition (the **Base Case**) where it stops calling itself.

**The Call Stack**
Every time a function is called, the system sets aside memory (a stack frame) for its local variables. When function A calls function B, function A is paused and pushed onto the stack. When B finishes, A is popped and resumes.

Because recursion uses LIFO (Last In, First Out) ordering, the stack naturally unwinds in reverse order of the calls.


````python
def infinite_recursion():
    print("Going down!")
    infinite_recursion()  # No base case = RecursionError (Stack Overflow)

# Uncomment to crash:
# infinite_recursion()
````

---
## 2. The Faith / Trust Principle

Stop trying to trace every recursive step in your head. Instead:
1. **Trust** that your function works for `n-1`.
2. Build the solution for `n` using the result of `n-1`.

If the base case is correct, and the link between `n` and `n-1` is correct, mathematical induction guarantees the whole function is correct.


---
## 3. Base Cases & Identity Elements

The Base Case is the anchor.
- For summation: Return `0` (Identity element of addition).
- For product: Return `1` (Identity element of multiplication).
- For min/max: Return `infinity` or `-infinity`.
- For strings/arrays: Return when `len == 0` or `left >= right`.


---
## 4. Preorder, Postorder & Backtracking

**Preorder:** Print before the recursive call.
**Postorder:** Print after the recursive call returns.
**Backtracking:** Modify state, recursively call, then revert state.


````python
def print_pre_post(n):
    if n == 0:
        print("Base case hit!")
        return
    print(f"Preorder: {n}")
    print_pre_post(n - 1)
    print(f"Postorder: {n}")

print_pre_post(3)
````

# Problem 1: Print Name N Times

## Intuition
We want to print a string N times. We can do this by keeping a counter `i` that goes from 1 to N.

## Recursive Thinking
1. Base Case: If `i > N`, stop.
2. Work: Print the name.
3. Recursive call: Pass `i + 1` to the next call.

## Dry Run
For N=3, Name='Maitraiyee':
1. `f(1, 3)` -> Prints Maitraiyee -> Calls `f(2, 3)`
2. `f(2, 3)` -> Prints Maitraiyee -> Calls `f(3, 3)`
3. `f(3, 3)` -> Prints Maitraiyee -> Calls `f(4, 3)`
4. `f(4, 3)` -> `4 > 3`, triggers Base Case, returns.

## Time & Space Complexity
**Time:** `O(N)` because we make N function calls.
**Space:** `O(N)` because the maximum depth of the call stack is N.


````python
def print_name(i, n, name):
    if i > n:
        return
    print(name)
    print_name(i + 1, n, name)

print_name(1, 3, 'Recursion is fun!')
````

## Pattern: Parameterized Recursion (Preorder)
## Interview Notes: Extremely basic. Make sure to clearly state Space Complexity is O(N) due to the call stack.

# Problem 2: Reverse Array

## Intuition
To reverse an array, we can swap the first and last elements, then move inward and recursively swap the rest of the array.

## Recursive Thinking
Instead of a loop moving `left` and `right` pointers, we pass `left` and `right` as parameters.
Base Case: When `left >= right`, all elements are swapped. Stop.

## Dry Run
`arr = [1, 2, 3, 4, 5]`
1. `reverse(arr, 0, 4)` -> Swap arr[0] and arr[4] -> `[5, 2, 3, 4, 1]` -> Calls `reverse(arr, 1, 3)`
2. `reverse(arr, 1, 3)` -> Swap arr[1] and arr[3] -> `[5, 4, 3, 2, 1]` -> Calls `reverse(arr, 2, 2)`
3. `reverse(arr, 2, 2)` -> `left (2) >= right (2)`, Base Case! Returns.

## Time & Space Complexity
**Time:** `O(N/2)` = `O(N)`
**Space:** `O(N/2)` = `O(N)` due to call stack.


````python
def reverse_array_two_pointers(arr, left, right):
    if left >= right:
        return arr
    arr[left], arr[right] = arr[right], arr[left]
    return reverse_array_two_pointers(arr, left + 1, right - 1)

def reverse_array_one_pointer(arr, i):
    n = len(arr)
    if i >= n // 2:
        return arr
    arr[i], arr[n - i - 1] = arr[n - i - 1], arr[i]
    return reverse_array_one_pointer(arr, i + 1)

arr1 = [1, 2, 3, 4, 5]
arr2 = [1, 2, 3, 4, 5]
print(f"Two pointers: {reverse_array_two_pointers(arr1, 0, len(arr1) - 1)}")
print(f"One pointer: {reverse_array_one_pointer(arr2, 0)}")
````

## Pattern: Parameterized Recursion with State Modification
## Interview Notes: The single pointer version is often asked to test if you can derive the right pointer (`N - i - 1`) from the left pointer `i`.

# Problem 3: Check Palindrome

## Intuition
A string is a palindrome if the first and last characters match, AND the inner substring is also a palindrome.

## Recursive Thinking
Base Case: If `left >= right`, we have successfully checked all pairs. Return `True`.
Failure Case: If `s[left] != s[right]`, return `False` immediately.
Recursive step: Ask the smaller subproblem if it's a palindrome.

## Time & Space Complexity
**Time:** `O(N/2)` = `O(N)`
**Space:** `O(N/2)` = `O(N)` auxiliary stack space.


````python
def is_palindrome(s, left, right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False
    return is_palindrome(s, left + 1, right - 1)

s1 = "madam"
s2 = "hello"
print(f"{s1}: {is_palindrome(s1, 0, len(s1)-1)}")
print(f"{s2}: {is_palindrome(s2, 0, len(s2)-1)}")
````

# Problem 4: Fibonacci

## Intuition
The Nth Fibonacci number is the sum of the (N-1)th and (N-2)th Fibonacci numbers.

## Recursive Thinking (Multiple Recursion)
Base Case: `fib(0) = 0`, `fib(1) = 1`. So if `n <= 1`, return `n`.
Recursive Step: Return `fib(n-1) + fib(n-2)`.

## Dry Run & Recursion Tree
For `fib(4)`:
```text
                          fib(4)
                      /            \
                fib(3)              fib(2)
               /      \            /      \
          fib(2)     fib(1)    fib(1)    fib(0)
         /      \
     fib(1)    fib(0)
```

## Time & Space Complexity
**Time:** `O(2^N)`. Every node splits into two. Exponential time! Very slow.
**Space:** `O(N)`. The maximum depth of the recursion tree is N (the path `fib(4) -> fib(3) -> fib(2) -> fib(1)`).

## Motivation for Dynamic Programming
Look at the tree above. `fib(2)` is computed TWICE entirely independently. For larger numbers like `fib(50)`, these *overlapping subproblems* are computed millions of times. By caching results (DP), time becomes `O(N)`.


````python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print(f"Fibonacci 6: {fib(6)}")
````

## Pattern: Multiple Recursion
## Interview Notes: Always mention that naive recursion for Fibonacci is bad practice due to overlapping subproblems, and immediately suggest Memoization.
