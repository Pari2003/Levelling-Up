Sets are like list but are unordered and can not contain duplicate values but unique values only. if u try to add duplicate values, it will be ignored

```python
from typing import Set
def list_to_set(nums: List[int])-> Set[int]:
  my_set= set()
  for i in nums:
    my_set.add(i)
  return my_set
print(list_to_set([1,2,3,4,5]))
```

```python
#or use this method to convert list to set
def list_to_set(nums: List[int])-> Set[int]:
  return set(nums)
print(list_to_set([1,2,3,4,5,5,5]))
```

```python
def count_unique_words(words: List[str])-> int:
  if len(words)>0:
    count= 0
    my_set= set(words)
    for i in my_set:
      count+=1
    return count
  return 0
print(count_unique_words(["hello", "world","world"]))
```

```python
#or we can use len function:
def count_unique_words(words: List[str])-> int:
  if len(words)>0:
    my_set= set(words)
    return len(my_set)
  return 0
print(count_unique_words(["hello", "world","world"]))
```

```python
def contains_duplicate(words:List[str]) ->bool:
  my_set= set(words)
  return len(my_set)!= len(words)
print(contains_duplicate(["hell0","world"]))
print(contains_duplicate(["hell0","world","world"]))
```

Dictionary: can not contain duplicate keys like sets

keys need to be unique, but values need not be unique so multiple keys can point to same value

item() function will give us both the key and value pair in the dictionary

```python
from typing import Dict
def create_dict(name: str, age: int)-> Dict[str,int]:
  my_dict={}
  my_dict[name]= age
  return my_dict
def list_to_dict(words:List[str])-> Dict[str,int]:
  my_dict= {}
  for i in words:
    my_dict[i]= words.index(i)
  return my_dict
print(create_dict("pari",22))
print(list_to_dict("pari"))
```

```python
#say u were given multiple words in list, and have to give index values to each word
def list_to_dict(words:List[str])-> Dict[str,int]:
  my_dict={}
  for i in range(len(words)):
    w= words[i]
    my_dict[w]= i
  return my_dict
print(list_to_dict(["alice", "pari", "neo"]))
```

```python
def get_dict_keys(age_dict: Dict[str,int])-> List[str]:
  my_list=[]
  for key in age_dict:
    my_list.append(key)
  return my_list
def get_dict_values(age_dict:Dict[str,int])-> List[int]:
  my_list=[]
  for value in age_dict:
    age= age_dict[value]
    my_list.append(age)
  return my_list

print(get_dict_keys({"pari":22, "alice": 35}))
print(get_dict_values({"pari":22, "alice": 35}))
```

```python
#imp: count the occurances of elements in a list or str
#here we need to put if statement as first the dictionary is empty and we are incrementing it but nothing exists
#like count+=1 is basically count=count +1, so it will give key error othewise
def count_char(word:str)->Dict[str,int]:
  my_dict={}
  for char in word:
    if char not in my_dict:
      my_dict[char]=0
    my_dict[char]+=1
  return my_dict
print(count_char("hello"))
```

```python
def remove_keys(my_dict: Dict[str,int], keys: List[str])-> Dict[str,int]:
  for i in keys:
    my_dict.pop(i,0)
  return my_dict
print(remove_keys({"a":1,"b":2,"c":3},["a"]))
print(remove_keys({"a":1,"b":2,"c":3},["d"]))
```

Reading input from the console

but we need to be careful that if we are accepting integer and user inputs string then that would give error

```python
def read_and_print_input()-> None:
  op= input("Reading one line of text into a string: ")
  print(op)
read_and_print_input()
read_and_print_input()
```

```python
def read_integer()->int:
  op=int(input())
  return op
def read_float()->float:
  op=float(input())
  return op
read_integer()
read_float()
```

```python
def read_integer()-> List[int]:
  line=input()
  list_of_strings= line.split(",")
  list_of_int= []
  for s in list_of_strings:
    list_of_int.append(int(s))
  return list_of_int
print(read_integer())
```

```python
def add_two_numbers()-> int:
  line= input()
  numbers= line.split(",")
  return int(numbers[0])+ int(numbers[1])
print(add_two_numbers())
```

Try Except

```python
def divide_numbers(a:int,b:int)-> None:
  try:
    return a/b
  except:
    print("An error occurred!")
print(divide_numbers(2,3))
print(divide_numbers(2,0))
```

```python
def divide_numbers(a:str, b:str)-> None:
  try:
    a_int= int(a)
    b_int=int(b)
    return a_int/b_int
  except Exception as e:
    print("an error occurred:", e)
print(divide_numbers("1","2"))
print(divide_numbers("not a number","2"))
print(divide_numbers("1","0"))
```

```python
#multiple except blocks
#we should only have separate blocks for cases we want to give some custom msg and still should have general error block too
try:
  num1= int(a)
  num2= int(b)
  result= num1/num2
except ValueError:
  print("invalid value")
except ZeroDivisionError:
  print("dividing by zero")
except Exception as e:
  print("some exception has occured")
```
