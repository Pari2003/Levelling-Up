## 2. Data Types & Dynamic Typing

### Dynamic Typing

dynamic typing is not a good idea

Python is **dynamically typed** — a variable can be reassigned to a value of a different type at any time. While this makes Python flexible, it can lead to bugs that are hard to catch. In statically-typed languages (like Java or C++), the type is fixed at declaration.

> **Interview Note:** Dynamic typing means type errors only surface at runtime, not at compile time. This is why Python's type hints (`x: int = 5`) are encouraged for larger codebases.

```python
var= 10
print(type(var))
var= 10.0
print(type(var))
var= True
print(type(var))
var= "10"
print(type(var))
var=[10]
print(type(var))
```

### Type Casting

type_casting but you can not convert some types like string to int which will give type_error but works if we put int in "" but only if what we are type casting to integer is an integer in string

Type casting converts one data type into another using built-in functions like `int()`, `float()`, `str()`, `bool()`.

> **Edge Cases That Will Crash:**
> - `int("hello")` → `ValueError` (not a number)
> - `int("3.14")` → `ValueError` (has a decimal; use `int(float("3.14"))` instead)
> - `int("")` → `ValueError` (empty string)
>
> ✅ **What Works:** `int("10")`, `float("3.14")`, `str(42)`, `bool(0)` → `False`

```python
var= "10"
print(var)
print(type(var))
var= int(var)
print(var)
print(type(var))
```

## 3. Operators

### Order of Operations (PEMDAS)

Order of operations= PEMDAS (Parentheses, Exponents, Multiplication, Division, Addition, Subtraction)

Python follows standard mathematical precedence. Use parentheses `()` to make your intended order explicit and readable.

> **Common Pitfall:** Division `/` in Python 3 always returns a `float`, even for `4/2` which returns `2.0`, not `2`.

```python
a,b,c= 2,2,0.5
#1
print(a+b+c)
#2
print(0-(a+b+c))
#3
print(a*b*c)
#4
print((a+b+c)/(a*b*c))
```

order: exp->fd->mod

2.   % is modulus that will give the reminder


1.   ** is exponent

> **Edge Cases:**
> - `7 // 2` = `3` (rounds down), but `-7 // 2` = `-4` (rounds towards negative infinity, NOT towards zero!)
> - `7 % 2` = `1`, but `-7 % 2` = `1` in Python (not `-1` like in C/Java)
> - `0 ** 0` = `1` in Python (mathematically debatable, but Python defines it this way)

## 4. Logical Operators

### Logical OR

Logical OR

The `or` operator returns `True` if **at least one** operand is `True`. It uses **short-circuit evaluation** — if the first operand is `True`, Python doesn't even check the second one.

> **Interview Insight:** `or` doesn't just return `True`/`False` — it returns the first truthy value, or the last value if all are falsy. e.g., `0 or "hello"` returns `"hello"`.

### Keywords in Python

A keyword in python are english words reserved for values that can not be used as a variable. eg- or, True, False

Python has 35 reserved keywords (as of Python 3.12). You can see them all with `import keyword; print(keyword.kwlist)`.

```python
a,b,c,d= False, False, True, True
print(a or b)
print(b or c)
print(c or d)
print(a or b or c or d)
```

### Logical AND

Logical AND

The `and` operator returns `True` only if **both** operands are `True`. Also uses short-circuit evaluation — if the first operand is `False`, the second is never checked.

> **Interview Insight:** `and` returns the first falsy value, or the last value if all are truthy. e.g., `1 and 2 and 3` returns `3`.

```python
a,b,c,d = False, False, True, True
print(a and b)
print(b and c)
print(c and d)
print( a and b and c and d)
```

### Logical NOT (Negation)

Logical NOT(negation)

The `not` operator flips `True` to `False` and vice versa. It has the highest precedence among logical operators: `not` > `and` > `or`.

```python
a,b = True, False
print(not(a and b))
```

```python
a,b,c = False, False, True
print(not a)
print(not c)
print(not(a and b))
print(not(b or c))
```

### Elif (Else If)

Elif: allows us to check multiple conditions in order

Use `elif` to chain multiple conditions. Python evaluates them **top to bottom** and executes **only the first match**.

elif has if because it has a condition that will only execute if "if statement did not execute

> **Tip:** `elif` is short for "else if". It's NOT a separate `if` — it's part of the same chain.

only one condition will be true. even if multiple conditons are found to be true, it will end when it encounters the first true condition.

### Classic Interview Problem: FizzBuzz

FizzBuzz is one of the most common beginner interview questions. The rules:
- Print `"fizz"` if divisible by 3
- Print `"buzz"` if divisible by 5
- Print `"fizzbuzz"` if divisible by both
- Otherwise, print the number itself

> **Edge Case in this code:** It starts from 0, which is divisible by both 3 and 5, so the first output is `"fizzbuzz"`. Many interviewers expect the range to start from 1.

```python
for i in range(100):
  result= ""
  if i%3==0:
    result+= "fizz"
  if i%5==0:
    result+= "buzz"
  if i%3!=0 and i%5!=0:
    result+= str(i)
  print(result)
```

### Nested Loops

Nested Loop

A loop inside another loop. The inner loop runs completely for each iteration of the outer loop.

> **Time Complexity:** Nested loops generally have O(n²) time complexity. This is important for interview problems — nested loops over large inputs can be slow.

break: exits loop immediately

continue: skips remaining code in that iteration of the loop and will move to next iteration

### Control Flow: `break`, `continue`, `pass`

Control Flow(few keywords that can control the execution of the loop from inside the loop): break, continue, pass

These three keywords alter the normal flow of loop execution:

```python
**`break`:** Exits the loop immediately, skipping all remaining iterations.

break: exits loop immediately

> 💡 **Use Case:** Searching for an element in a list — once found, no need to check the rest.
```

```python
**`continue`:** Skips the rest of the current iteration and jumps to the next one.

continue: skips remaining code in that iteration of the loop and will move to next iteration

> 💡 **Use Case:** Skip invalid or unwanted items during iteration (e.g., skip negative numbers).
```

```python
**`pass`:** Does absolutely nothing. Used as a placeholder for empty code blocks (since Python doesn't allow empty blocks).

pass: does nothing and can be used outside of the loop. it acts like a placeholders since we can not have empty loop

> 💡 **Use Case:** Stub out functions or classes you plan to implement later: `def todo(): pass`
```

```python
#here it skipped 3 because continue will literally end the iteration and start with the next one when encountered
#and all after 6 including 6 will not be printed because it will end the loop itself
for i in range(1,8):
  if i == 3:
    continue
  elif i == 6:
    break
  print(i)
```

### Sets

Sets are like list but are unordered and can not contain duplicate values but unique values only. if u try to add duplicate values, it will be ignored

Sets use curly braces `{}` or the `set()` constructor. They are unordered collections of unique elements.

> **Interview Superpower:** Set lookups (`x in my_set`) are **O(1)** time complexity. They are perfect for removing duplicates from a list or checking for existence fast.

```python
from typing import Set
def list_to_set(nums: List[int])-> Set[int]:
  my_set= set()
  for i in nums:
    my_set.add(i)
  return my_set
print(list_to_set([1,2,3,4,5]))
```

```python
def count_unique_words(words: List[str])-> int:
  if len(words)>0:
    count= 0
    my_set= set(words)
    for i in my_set:
      count+=1
    return count
  return 0
print(count_unique_words(["hello", "world","world"]))
```

### Dictionaries (Hash Maps)

Dictionary: can not contain duplicate keys like sets

Dictionaries store data in **key-value pairs**, allowing fast O(1) lookups by key. They use curly braces `{}`.

> **Interview Tip:** Dictionaries in Python are essentially Hash Maps. Whenever a problem asks you to "count frequencies" or "find a pair of numbers", a dictionary is usually the answer (e.g., Two Sum).

keys need to be unique, but values need not be unique so multiple keys can point to same value

> **Key Constraints:** Keys must be **immutable** (strings, integers, tuples). You cannot use a list or another dictionary as a dictionary key!

item() function will give us both the key and value pair in the dictionary

> **Iteration Methods:**
> - `dict.keys()`: Iterates keys
> - `dict.values()`: Iterates values
> - `dict.items()`: Iterates (key, value) tuples

```python
from typing import Dict
def create_dict(name: str, age: int)-> Dict[str,int]:
  my_dict={}
  my_dict[name]= age
  return my_dict
def list_to_dict(words:List[str])-> Dict[str,int]:
  my_dict= {}
  for i in words:
    my_dict[i]= words.index(i)
  return my_dict
print(create_dict("pari",22))
print(list_to_dict("pari"))
```

```python
#say u were given multiple words in list, and have to give index values to each word
def list_to_dict(words:List[str])-> Dict[str,int]:
  my_dict={}
  for i in range(len(words)):
    w= words[i]
    my_dict[w]= i
  return my_dict
print(list_to_dict(["alice", "pari", "neo"]))
```

## 11. Error Handling (Try / Except)

### Try / Except Block

Try Except

When Python encounters an error (Exception), it normally crashes the program. Using a `try...except` block allows you to "catch" the error and handle it gracefully without crashing.

```python
def divide_numbers(a:int,b:int)-> None:
  try:
    return a/b
  except:
    print("An error occurred!")
print(divide_numbers(2,3))
print(divide_numbers(2,0))
```

### Catching Specific Errors

error catching

It's bad practice to use a "bare except" (`except:`) because it catches ALL errors, even ones you didn't anticipate (like `KeyboardInterrupt` if the user tries to exit).

> **Best Practice:** Catch specific errors like `ValueError`, `ZeroDivisionError`, `TypeError`, etc., so you can handle different failures in different ways.

```python
try:
  result= 10/0
except Exception as error:
  print("Error:", error)
```

```python
def divide_numbers(a:str, b:str)-> None:
  try:
    a_int= int(a)
    b_int=int(b)
    return a_int/b_int
  except Exception as e:
    print("an error occurred:", e)
print(divide_numbers("1","2"))
print(divide_numbers("not a number","2"))
print(divide_numbers("1","0"))
```

```python
#multiple except blocks
#we should only have separate blocks for cases we want to give some custom msg and still should have general error block too
try:
  num1= int(a)
  num2= int(b)
  result= num1/num2
except ValueError:
  print("invalid value")
except ZeroDivisionError:
  print("dividing by zero")
except Exception as e:
  print("some exception has occured")
```
